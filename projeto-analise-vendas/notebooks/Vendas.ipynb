```python
import pandas as pd

# Carregando os dados
# O arquivo 'supermarket_sales.csv' não foi encontrado no caminho especificado.
# Por favor, verifique se o arquivo foi carregado para o ambiente do Colab
# ou ajuste o caminho para onde o arquivo está localizado.
df = pd.read_csv('/SuperMarket Analysis.csv')

# Visualizando as primeiras linhas e informações gerais
print(df.head())
print(df.info())

# Tratando a coluna de data (essencial para analistas!)
df['Date'] = pd.to_datetime(df['Date'])

# Criando uma coluna de lucro bruto para análise
df['Gross_Profit'] = df['Sales'] - df['cogs']

print("Dados prontos para análise!")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurando o estilo visual
plt.figure(figsize=(12,6))
sns.set_theme(style="whitegrid")

# Criando o gráfico de faturamento por categoria
vendas_por_categoria = df.groupby('Product line')['Sales'].sum().sort_values(ascending=False)
vendas_por_categoria.plot(kind='bar', color='skyblue')

plt.title('Faturamento Total por Linha de Produto', fontsize=15)
plt.xlabel('Categoria', fontsize=12)
plt.ylabel('Total de Vendas (USD)', fontsize=12)
plt.xticks(rotation=45)
plt.savefig('vendas_por_categoria.png', bbox_inches='tight', dpi=300)
plt.show()
```
